```python
from ultralytics import settings

# Update the default download directory
settings.update({"weights_dir": "/workspace/models"})

from ultralytics import YOLOE

# Initialize a YOLOE model
model = YOLOE("yoloe-26m-seg.pt")  # or yoloe-26n/s/m/l-seg.pt for different sizes
```

```python
local_image_path = '/workspace/projects/Grounded-Segment-Anything/assets/inpaint_demo.jpg'
```

```python
# Set text prompt to detect person and bus. You only need to do this once after you load the model.
model.set_classes(["dog", "bench"])

# Run detection on the given image
results = model.predict(local_image_path)
```

```python
# Show results
results[0].show()
```
